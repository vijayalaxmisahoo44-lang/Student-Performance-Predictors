```python
# Day 9 - Python Tuples

student = ("Rahul", 20, "BCA", 75)

print("Student Information:", student)
```

```python
scores = (60, 75, 82, 68, 90)

print("Scores:", scores)
print("Highest Score:", max(scores))
print("Lowest Score:", min(scores))
print("Total Score:", sum(scores))
```

```python
for score in scores:
    print("Student Score:", score)
```

```python
print("Number of Score:",len(scores))
```
