```python
import pandas as pd
import matplotlib.pyplot as plt

# Create student dataset
data = {
    "Name": ["Rahul", "Priya", "Aman", "Sneha", "Riya", "Karan"],
    "Study_Hours": [2, 3, 5, 1, 4, 6],
    "Marks": [50, 60, 85, 40, 75, 95],
    "Attendance": [70, 80, 90, 60, 85, 95]
}

df = pd.DataFrame(data)

# Display student data
print("Student Performance Data:")
print(df)

# Calculate average marks
print("\nAverage Marks:")
print(df["Marks"].mean())

# Find the highest marks
print("\nHighest Marks:")
print(df["Marks"].max())

# Find the top-performing student
print("\nTop Performing Student:")
print(df.loc[df["Marks"].idxmax()])

# Bar chart of student marks
plt.bar(df["Name"], df["Marks"])
plt.title("Student Marks Analysis")
plt.xlabel("Student Name")
plt.ylabel("Marks")
plt.show()

# Scatter plot of study hours and marks
plt.scatter(df["Study_Hours"], df["Marks"])
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.show()
```
