```python
import pandas as pd
import matplotlib.pyplot as plt

#Create student dataset

data = {
"Student": ["Aman", "Riya", "Rahul", "Priya", "Neha"],
"Study_Hours": [2, 4, 5, 7, 9],
"Attendance": [60, 70, 80, 85, 95],
"Marks": [35, 45, 60, 75, 90]
}

df = pd.DataFrame(data)

#Display dataset

print("STUDENT PERFORMANCE DASHBOARD")
print(df)

#Calculate statistics

print("\nTotal Students:", len(df))
print("Average Marks:", df["Marks"].mean())
print("Highest Marks:", df["Marks"].max())
print("Lowest Marks:", df["Marks"].min())

#Classify performance

def performance(marks):
  if marks >= 75:
    return "Excellent"
  elif marks >= 50:
    return "Average"
  else:
    return "Needs Improvement"

df["Performance"] = df["Marks"].apply(performance)

print("\nPerformance Analysis:")
print(df[["Student", "Marks", "Performance"]])

#Plot student marks

plt.figure(figsize=(8, 5))
plt.bar(df["Student"], df["Marks"])
plt.title("Student Marks Comparison")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.show()

#Plot study hours vs marks

plt.figure(figsize=(8, 5))
plt.plot(df["Study_Hours"], df["Marks"], marker="o")
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.grid(True)
plt.show()
```
