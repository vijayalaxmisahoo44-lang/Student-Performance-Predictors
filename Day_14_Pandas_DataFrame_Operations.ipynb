```python
import pandas as pd

data = {
    "Name": ["Rahul", "Priya", "Aman", "Sneha", "Riya"],
    "Marks": [85, 78, 92, 67, 88],
    "Attendance": [90, 85, 95, 75, 88]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

# Select a column
print("\nNames:")
print(df["Name"])

# Add a new column
df["Result"] = ["Pass", "Pass", "Pass", "Pass", "Pass"]

print("\nAfter adding Result:")
print(df)

# Filter students with marks greater than 80
print("\nStudents with marks above 80:")
print(df[df["Marks"] > 80])
```
