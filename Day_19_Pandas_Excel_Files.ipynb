```python
import pandas as pd

# Create student data
data = {
    "Name": ["Rahul", "Priya", "Aman", "Sneha", "Riya"],
    "Branch": ["CSE", "CSE", "BCA", "BCA", "CSE"],
    "Marks": [85, 78, 92, 67, 88]
}

df = pd.DataFrame(data)

print("Original Student Data:")
print(df)

# Save data to Excel file
df.to_excel("student_data.xlsx", index=False)

print("\nExcel file created successfully!")

# Read data from Excel file
new_df = pd.read_excel("student_data.xlsx")

print("\nData read from Excel:")
print(new_df)

# Display students scoring above 80
print("\nStudents scoring above 80:")
print(new_df[new_df["Marks"] > 80])
```
