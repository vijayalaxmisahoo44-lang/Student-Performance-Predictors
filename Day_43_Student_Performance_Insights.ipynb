```python
import pandas as pd
import matplotlib.pyplot as plt

# Create student dataset
data = {
    "Student": ["Aman", "Riya", "Rahul", "Priya", "Neha"],
    "Study_Hours": [2, 4, 5, 7, 9],
    "Attendance": [60, 70, 80, 85, 95],
    "Marks": [35, 45, 60, 75, 90]
}

df = pd.DataFrame(data)

# Analyze student performance
def get_insights(row):
    if row["Marks"] >= 75:
        return "Excellent Performance"
    elif row["Marks"] >= 50:
        return "Good! Keep Practicing"
    else:
        return "Need More Study and Practice"

df["Insights"] = df.apply(get_insights, axis=1)

# Display results
print("STUDENT PERFORMANCE INSIGHTS")
print(df)

# Generate recommendations
print("\nSTUDENT RECOMMENDATIONS")

for index, row in df.iterrows():
    print("\nStudent:", row["Student"])

    if row["Study_Hours"] < 4:
        print("Recommendation: Increase daily study hours.")

    if row["Attendance"] < 75:
        print("Recommendation: Improve class attendance.")

    if row["Marks"] < 50:
        print("Recommendation: Practice more and revise regularly.")

    if row["Marks"] >= 75:
        print("Recommendation: Maintain your excellent performance.")

# Visualize student performance
plt.figure(figsize=(8, 5))
plt.bar(df["Student"], df["Marks"])
plt.title("Student Performance Analysis")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()

# Save insights report
df.to_csv("Day_43_Student_Insights.csv", index=False)

print("\nInsights report saved successfully!")
```
