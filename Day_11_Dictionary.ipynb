```python
# Day 11 - Python Dictionary

student = {
    "name": "Rahul",
    "age": 20,
    "branch": "BCA",
    "study_hours": 6,
    "previous_score": 75
}

print("Student Information:")
print(student)
```

```python
print("Name:", student["name"])
print("Branch:", student["branch"])
print("Study Hours:", student["study_hours"])
print("Previous Score:", student["previous_score"])
```

```python
student["attendance"] = 85
print("After Adding Attendance:", student)

student["study_hours"] = 7
print("Updated Study Hours:", student["study_hours"])
```

```python
student["final_score"] = 82

print("Final Score:", student["final_score"])
```

```python
for key, value in student.items():
    print(key, ":", value)
```
