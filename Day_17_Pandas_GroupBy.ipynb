```python
import pandas as pd

data = {
    "Name": ["Rahul", "Priya", "Aman", "Sneha", "Riya", "Karan"],
    "Branch": ["CSE", "CSE", "BCA", "BCA", "CSE", "BCA"],
    "Marks": [85, 78, 92, 67, 88, 75]
}

df = pd.DataFrame(data)

print("Student Data:")
print(df)

# Average marks by branch
print("\nAverage Marks by Branch:")
print(df.groupby("Branch")["Marks"].mean())

# Highest marks by branch
print("\nHighest Marks by Branch:")
print(df.groupby("Branch")["Marks"].max())

# Lowest marks by branch
print("\nLowest Marks by Branch:")
print(df.groupby("Branch")["Marks"].min())

# Number of students in each branch
print("\nNumber of Students in Each Branch:")
print(df.groupby("Branch")["Name"].count())
```
