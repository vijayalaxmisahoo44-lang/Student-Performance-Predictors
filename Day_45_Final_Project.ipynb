```python
import pandas as pd
import matplotlib.pyplot as plt

# Student Dataset
data = {
    "Student": ["Aman", "Riya", "Rahul", "Priya", "Neha"],
    "Study_Hours": [2, 4, 5, 7, 9],
    "Attendance": [60, 70, 80, 85, 95],
    "Marks": [35, 45, 60, 75, 90]
}

df = pd.DataFrame(data)

# Performance Classification
def performance(marks):
    if marks >= 75:
        return "Excellent"
    elif marks >= 50:
        return "Average"
    else:
        return "Needs Improvement"

df["Performance"] = df["Marks"].apply(performance)

# Display Final Report
print("STUDENT PERFORMANCE REPORT")
print(df)

# Average Marks
print("\nAverage Marks:", df["Marks"].mean())

# Topper
topper = df.loc[df["Marks"].idxmax()]
print("\nTopper:", topper["Student"])
print("Highest Marks:", topper["Marks"])

# Visualization
plt.bar(df["Student"], df["Marks"])
plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Performance Analysis")
plt.show()

# Save Report
df.to_csv("Final_Student_Performance.csv", index=False)

print("\nFinal Project Completed Successfully!")
```
