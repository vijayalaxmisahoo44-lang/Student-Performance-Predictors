```python
# Day 12 - Python Practice

student = {
    "name": "Rahul",
    "study_hours": 6,
    "attendance": 85,
    "previous_score": 72
}

print("Student Name:", student["name"])
print("Study Hours:", student["study_hours"])
print("Attendance:", student["attendance"])
print("Previous Score:", student["previous_score"])
```

```python
final_score = 80

average_score = (student["previous_score"] + final_score) / 2

print("Final Score:", final_score)
print("Average Score:", average_score)
```

```python
if final_score >= 75:
    performance = "Excellent"
elif final_score >= 60:
    performance = "Good"
elif final_score >= 45:
    performance = "Average"
else:
    performance = "Needs Improvement"

print("Performance:", performance)
```

```python
scores = [60, 70, 75, 80, 85]

total = 0

for score in scores:
    total += score

print("Scores:", scores)
print("Total:", total)
print("Highest:", max(scores))
print("Lowest:", min(scores))
```

```python
print("----- Student Performance Report -----")
print("Name:", student["name"])
print("Study Hours:", student["study_hours"])
print("Attendance:", student["attendance"], "%")
print("Previous Score:", student["previous_score"])
print("Final Score:", final_score)
print("Average Score:", average_score)
print("Performance:", performance)
```
