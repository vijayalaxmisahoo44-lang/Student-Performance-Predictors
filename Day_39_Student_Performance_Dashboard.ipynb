```python
import pandas as pd
import matplotlib.pyplot as plt

# Create student dataset
data = {
    "Student": ["A", "B", "C", "D", "E"],
    "Study_Hours": [2, 4, 5, 7, 9],
    "Marks": [35, 45, 60, 75, 90]
}

df = pd.DataFrame(data)

# Display dashboard title
print("STUDENT PERFORMANCE DASHBOARD")
print("=" * 35)

# Display student data
print("\nStudent Details:")
print(df.to_string(index=False))

# Calculate statistics
average_marks = df["Marks"].mean()
highest_marks = df["Marks"].max()
lowest_marks = df["Marks"].min()

print("\nAverage Marks:", round(average_marks, 2))
print("Highest Marks:", highest_marks)
print("Lowest Marks:", lowest_marks)

# Create performance chart
plt.figure(figsize=(8, 5))
plt.bar(df["Student"], df["Marks"])
plt.axhline(average_marks, linestyle="--", label="Average Marks")
plt.title("Student Performance Dashboard")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.legend()
plt.show()

# Study hours vs marks
plt.figure(figsize=(8, 5))
plt.plot(df["Study_Hours"], df["Marks"], marker="o")
plt.title("Study Hours vs Student Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.grid(True)
plt.show()
```
