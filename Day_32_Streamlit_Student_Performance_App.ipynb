```python
import pandas as pd
import matplotlib.pyplot as plt
import streamlit as st

st.title("Student Performance Predictor")

st.header("Student Details")

name = st.text_input("Enter Student Name")
study_hours = st.number_input(
    "Daily Study Hours",
    min_value=0.0,
    max_value=24.0,
    value=2.0
)

attendance = st.number_input(
    "Attendance Percentage",
    min_value=0,
    max_value=100,
    value=75
)

if st.button("Predict Performance"):
    predicted_marks = min(
        100,
        max(0, study_hours * 10 + attendance * 0.3)
    )

    st.success("Prediction Completed!")
    st.write("Student Name:", name)
    st.write("Predicted Marks:", round(predicted_marks, 2))

    if predicted_marks >= 75:
        st.write("Performance Level: Excellent")
    elif predicted_marks >= 50:
        st.write("Performance Level: Average")
    else:
        st.write("Performance Level: Needs Improvement")
```
