```python
# Day 7 - Python Loops

for i in range(1, 6):
    print("Student", i)
```

```python
study_hours = [3, 5, 6, 4, 7]

for hours in study_hours:
    print("Study Hours:", hours)
```

```python
study_hours = 1

while study_hours <= 5:
    print("Study Hours:", study_hours)
    study_hours += 1
```

```python
scores = [60, 70, 80, 90, 75]

total = 0

for score in scores:
    total += score

print("Total Score:", total)
```
