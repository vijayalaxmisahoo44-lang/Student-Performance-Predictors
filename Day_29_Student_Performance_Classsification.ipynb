```python
import pandas as pd

# Create student dataset
data = {
    "Name": ["Rahul", "Priya", "Aman", "Sneha", "Riya", "Karan"],
    "Marks": [85, 78, 92, 35, 55, 68]
}

df = pd.DataFrame(data)

print("Student Performance Data:")
print(df)

# Classify student performance
def classify_performance(marks):
    if marks >= 80:
        return "Excellent"
    elif marks >= 60:
        return "Good"
    elif marks >= 40:
        return "Average"
    else:
        return "Needs Improvement"

df["Performance"] = df["Marks"].apply(classify_performance)

print("\nStudent Performance Classification:")
print(df)

# Count students in each category
print("\nPerformance Category Count:")
print(df["Performance"].value_counts())
```
