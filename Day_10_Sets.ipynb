```python
# Day 10 - Python Sets

student_scores = {60, 75, 80, 75, 90}

print("Student Scores:", student_scores)
```

```python
student_scores.add(95)
print("After Adding:", student_scores)

student_scores.remove(60)
print("After Removing:", student_scores)
```

```python
set1 = {60, 70, 80, 90}
set2 = {70, 80, 85, 95}

print("Union:", set1 | set2)
print("Intersection:", set1 & set2)
```

```python
if 80 in student_scores:
    print("80 is present in the set")
else:
    print("80 is not present")
```

```python
print("Number of Scores:", len(student_scores))
```
