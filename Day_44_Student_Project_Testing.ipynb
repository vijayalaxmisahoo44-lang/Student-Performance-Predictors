```python
import pandas as pd

# Create student dataset
data = {
    "Student": ["Aman", "Riya", "Rahul", "Priya", "Neha"],
    "Study_Hours": [2, 4, 5, 7, 9],
    "Attendance": [60, 70, 80, 85, 95],
    "Marks": [35, 45, 60, 75, 90]
}

df = pd.DataFrame(data)

# Function to classify performance
def performance(marks):
    if marks >= 75:
        return "Excellent"
    elif marks >= 50:
        return "Average"
    else:
        return "Needs Improvement"

# Apply performance classification
df["Performance"] = df["Marks"].apply(performance)

# Test student records
print("STUDENT PERFORMANCE TESTING")
print(df)

# Test 1: Check total students
assert len(df) == 5
print("\nTest 1 Passed: Student count is correct.")

# Test 2: Check marks range
assert df["Marks"].between(0, 100).all()
print("Test 2 Passed: Marks are within valid range.")

# Test 3: Check missing values
assert not df.isnull().values.any()
print("Test 3 Passed: No missing values found.")

# Test 4: Check performance classification
assert performance(90) == "Excellent"
assert performance(60) == "Average"
assert performance(40) == "Needs Improvement"
print("Test 4 Passed: Performance classification is correct.")

# Save final report
df.to_csv("Final_Student_Report.csv", index=False)

print("\nAll tests completed successfully!")
print("Final report saved successfully.")
```
