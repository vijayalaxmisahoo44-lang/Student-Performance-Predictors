```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Student dataset
data = {
    "Study_Hours": [1, 2, 3, 4, 5, 6, 7, 8],
    "Marks": [35, 40, 50, 55, 65, 70, 80, 90]
}

df = pd.DataFrame(data)

print("Student Dataset:")
print(df)

# Input and output
X = df[["Study_Hours"]]
y = df["Marks"]

# Create the model
model = LinearRegression()

# Train the model
model.fit(X, y)

# Predict marks for 5 study hours
predicted_marks = model.predict(pd.DataFrame({"Study_Hours": [5]}))

print("\nPredicted marks for 5 study hours:")
print(round(predicted_marks[0], 2))

# Plot the data
plt.scatter(df["Study_Hours"], df["Marks"])
plt.plot(df["Study_Hours"], model.predict(X))

plt.title("Study Hours vs Student Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```
