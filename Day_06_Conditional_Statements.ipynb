```python
# Day 6 - Conditional Statements

final_score = 75

if final_score >= 60:
    print("Student Passed")
else:
    print("Student Failed")
```

```python
final_score = 75

if final_score >= 75:
    print("Excellent")
elif final_score >= 60:
    print("Good")
elif final_score >= 45:
    print("Average")
else:
    print("Needs Improvement")
```

```python
attendance = 85

if attendance >= 75:
    print("Attendance is Satisfactory")
else:
    print("Attendance is Low")
```

```python
study_hours = 6
final_score = 75

if study_hours >= 5 and final_score >= 60:
    print("Good Student Performance")
else:
    print("Student Needs Improvement")
```
