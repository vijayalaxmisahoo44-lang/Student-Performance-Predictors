```python
# Day 5 - Python Operators

study_hours = 6
previous_score = 70
final_score = 80

print("Addition:", previous_score + final_score)
print("Subtraction:", final_score - previous_score)
print("Multiplication:", study_hours * 2)
print("Division:", final_score / 2)
```

```python
print("Final score > Previous score:", final_score > previous_score)
print("Final score < Previous score:", final_score < previous_score)
print("Final score == Previous score:", final_score == previous_score)
```

```python
attendance = 85

print("Good Attendance:", attendance >= 75)
print("Good Performance:", final_score >= 60)

print("Eligible:", attendance >= 75 and final_score >= 60)
```

```python
study_hours = 5

study_hours += 2
print("Updated Study Hours:", study_hours)

study_hours -= 1
print("After Decrease:", study_hours)
```
