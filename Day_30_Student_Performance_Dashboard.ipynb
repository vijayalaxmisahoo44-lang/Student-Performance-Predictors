```python
import pandas as pd
import matplotlib.pyplot as plt

# Create student dataset
data = {
    "Name": ["Rahul", "Priya", "Aman", "Sneha", "Riya", "Karan"],
    "Marks": [85, 78, 92, 35, 55, 68],
    "Attendance": [90, 85, 95, 60, 75, 80]
}

df = pd.DataFrame(data)

# Display student data
print("STUDENT PERFORMANCE DASHBOARD")
print(df)

# Calculate average marks
average_marks = df["Marks"].mean()
print("\nAverage Marks:", round(average_marks, 2))

# Find top-performing student
top_student = df.loc[df["Marks"].idxmax()]
print("\nTop Performing Student:")
print(top_student)

# Count students who passed (marks >= 40)
passed = (df["Marks"] >= 40).sum()
failed = (df["Marks"] < 40).sum()

print("\nStudents Passed:", passed)
print("Students Failed:", failed)

# Create dashboard with three graphs
plt.figure(figsize=(15, 4))

# Graph 1: Student marks
plt.subplot(1, 3, 1)
plt.bar(df["Name"], df["Marks"])
plt.title("Student Marks")
plt.xlabel("Name")
plt.ylabel("Marks")
plt.xticks(rotation=45)

# Graph 2: Attendance
plt.subplot(1, 3, 2)
plt.bar(df["Name"], df["Attendance"])
plt.title("Student Attendance")
plt.xlabel("Name")
plt.ylabel("Attendance (%)")
plt.xticks(rotation=45)

# Graph 3: Marks distribution
plt.subplot(1, 3, 3)
plt.hist(df["Marks"], bins=5, edgecolor="black")
plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.tight_layout()
plt.show()
```
