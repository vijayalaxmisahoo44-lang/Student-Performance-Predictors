```python
import pandas as pd

# Create student data
data = {
    "Name": ["Rahul", "Priya", "Aman", "Sneha", "Riya"],
    "Branch": ["CSE", "CSE", "BCA", "BCA", "CSE"],
    "Marks": [85, 78, 92, 35, 88],
    "Attendance": [90, 85, 95, 60, 88]
}

df = pd.DataFrame(data)

print("Student Data:")
print(df)

# Calculate average marks
print("\nAverage Marks:")
print(df["Marks"].mean())

# Find highest marks
print("\nHighest Marks:")
print(df["Marks"].max())

# Find lowest marks
print("\nLowest Marks:")
print(df["Marks"].min())

# Find the highest-scoring student
print("\nTop Performing Student:")
print(df.loc[df["Marks"].idxmax()])

# Find students who scored below 40
print("\nStudents Needing Improvement:")
print(df[df["Marks"] < 40])

# Calculate average attendance
print("\nAverage Attendance:")
print(df["Attendance"].mean())
```
