```python
# Day 8 - Python Lists

student_names = ["Rahul", "Aman", "Priya", "Neha", "Riya"]

print("Student Names:", student_names)
```

```python
print("First Student:", student_names[0])
print("Third Student:", student_names[2])
```

```python
student_names.append("Karan")
print("After Adding:", student_names)

student_names.remove("Aman")
print("After Removing:", student_names)
```

```python
scores = [60, 75, 82, 68, 90]

print("Scores:", scores)
print("Highest Score:", max(scores))
print("Lowest Score:", min(scores))
print("Total Score:", sum(scores))
```

```python
for score in scores:
    print("Student Score:", score)
```
