```python
import pandas as pd

data = {
    "Name": ["Rahul", "Priya", "Aman", "Sneha", "Riya"],
    "Marks": [85, 78, 92, 67, 88],
    "Attendance": [90, 85, 95, 75, 88]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

# Sort by Marks in descending order
print("\nSorted by Marks:")
print(df.sort_values("Marks", ascending=False))

# Students with marks greater than 80
print("\nStudents with Marks > 80:")
print(df[df["Marks"] > 80])

# Students with attendance greater than 85
print("\nStudents with Attendance > 85:")
print(df[df["Attendance"] > 85])
```
